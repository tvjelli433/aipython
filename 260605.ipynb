```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn import datasets
iris = datasets.load_iris()
```

```python
print('iris 키:', iris.keys())
```

```python
print('데이터:\n', iris['data'][:5])
```

```python
print('특성 이름:', iris['feature_names'])
```

```python
print('타깃:\n', iris['target'][:5])
```

```python
print('타깃 이름:', iris['target_names'])
```

```python
print('데이터 크기:', iris['data'].shape)
print('타깃 크기:', iris['target'].shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(iris['data'], iris['target'], random_state=0)

print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3) #n_neighbors = 이웃개수 
```

```python
knn.fit(X_train, y_train)
```

```python
import numpy as np

X_new = np.array([[6, 3.5, 5, 1.5]])
```

```python
pred = knn.predict(X_new)
print(pred)
print(iris['target_names'][pred])
```

```python
y_pred = knn.predict(X_test)
print('테스트세트 예측값:\n', y_pred) #정답예측 (실제와 다를 수 있음)
print('테스트세트 실제타깃:\n', y_test) #실제정답
```

```python
print(f'훈련 세트 정확도: {knn.score(X_train, y_train):.3f}')
print(f'테스트 세트 정확도: {knn.score(X_test, y_test):.3f}') #.3f는 소수점 3자리까지
```

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

#데이터 수집하기
wine = datasets.load_wine()
print('wine 키:', wine.keys())
print('데이터:\n', wine['data'][:5])
print('특성이름:', wine['feature_names'])
print('타깃:\n', wine['target'][:5])
print('타깃이름:', wine['target_names'])
print('데이터 크기:', wine['data'].shape)
print('타깃 크기:', wine['data'].shape)
```

```python
#데이터 전처리
X_train, X_test, y_train, y_test = train_test_split(wine['data'], wine['target'], random_state=0)

print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

```python
#모델 생성하기
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
#모델 이용해서 예측하기
y_pred = knn.predict(X_test)
print('테스트세트 예측값:\n', y_pred)
print('테스트세트 실제타깃:\n', y_test)
```

```python
#모델 평가하기
print(f'훈련 세트 정확도: {knn.score(X_train, y_train):.3f}')
print(f'테스트 세트 정확도: {knn.score(X_test, y_test):.3f}')
```
